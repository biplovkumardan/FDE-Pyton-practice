# What is SDK?
## SDK stands for Software Development Kit
## SDK is collection of tools, libraries and functions provided by a company to help developers use to interact

## Suppose we want to communicate with Gemini AI from Python. We don't need to develop all the communication logic from scratch. Google provides a ready-made SDK called Google GenAI, We install that SDK using pip, import it into Python, create a client and call ready-made methods. That's the main purpose of an SDK: making developers' work easier by providing ready-made tools and functions.

### Google provides a Python SDK called google-genai (https://aistudio.google.com/docs/libraries)
#### --> pip install google-genai
#### --> from google import genai

### OpenAI provides an official Python SDK called openai
#### --> pip install openai
#### --> from openai import OpenAI

### Groq provides a Python SDK called groq
#### --> pip install groq
#### --> from groq import Groq

### Anthropc provides an official Python SDK called anthropic
#### --> pip install anthropic
#### --> from anthropic import Anthropic

```python
from google import genai
API_KEY = "Update you key"
client = genai.Client(api_key=API_KEY)

interaction = client.interactions.create(
    model="gemini-3.5-flash",
    input="Role of FDE in 5 sentences in the AI world?"
)

print(interaction.output_text)
```

```python
interaction1 = client.interactions.create(
    model="gemini-3.5-flash",
    input="A FDE can Build, Deploy and Ship. Can you explain these 3 keywords in 1 sentence for each?"
)

print(interaction1.output_text)
```

## With function

```python
from google import genai

client = genai.Client(api_key=API_KEY)

def ask_gemini(prompt):
    interaction = client.interactions.create(
        model = "gemini-3.5-flash",
        input = prompt 
    )
    return interaction.output_text

result = ask_gemini("What is your knowledge cut-off date?")
print(result)
```

```python
result1 = ask_gemini("what is python in one line?")
result2 = ask_gemini("what is java in one line?")
result3 = ask_gemini("what is aws in one line?")
print(result1)
print(result2)
print(result3)
```
