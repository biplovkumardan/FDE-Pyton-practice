# Indentation and Syntax

## Indentation means adding spaces begining line of the code

```python
if True:
print("hello")
# Here print() stmt is not indented( 4 space normally)
# The print stmt will be belong to if condition once indented
```

```python
# Within the blocks all print stmt indented same will belongs to that loop
if 1 in range(3):
    print("1 belongs to if block") # indented with 4 spaces
  print("2 not belong to if block") # indented with 2 spaces
```

```python
# Within the blocks all print stmt indented same will belongs to that loop
if 1 in range(3):
    print("1 belongs to if block") # indented with 4 spaces
    print("2 also belong to if block") # indented with 4 spaces
```

```python
# Within the blocks all print stmt indented same will belongs to that loop
if 1 in range(3):
    print("1 belongs to if block") # indented with 4 spaces
print("2 is outside if block") # Not indented
```

# Python Conditions if, elif, else, Nested condition

## Rule-1: Use colon (:) every if, else, elif stmt end
## Rule-2: Indentation mandatory
## Rule-3: elif, else cannot stand alone. It needs if stmt first.

```python
# example
marks = 75
if marks >= 90:
    print("Grade A")
if marks >= 75:
    print("Grade B")
if marks >= 60:
    print("Grade C")
else:
    print("fall")

# each condition is checked independently
# It is executing multiple matching blocks   
```

```python
# example
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("fall")

# Excute only one matching block 
```

## Nested Conditions
### Condition inside the condition

```python
#Nested condition
marks = 70
lab = 90
if marks >= 60:
    print("pass")
    if lab > 90:
        print("Eligible for 1st grade")
    else:
        print("Not eligible for 1st grade")
else:
    print("fall")
```
