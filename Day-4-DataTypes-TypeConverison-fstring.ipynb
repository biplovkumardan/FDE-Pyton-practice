## Conversion Type Process

### Convert a value from one data type to another data type

```python
## float to int conversion
a =2
b=3.5
c=a+b
print(c)
print(type(c))
# conversion to Int
c=int(a+b)
print(c)
print(type(c))
```

```python
# string to integer
age ="25"
age =int(age)
print(age)
print(type(age))
```

```python
age ="25" # string
age=list(age) # The list() function takes any iterable (like a string) and breaks it down item by item (or character by character) to build a new list.
print(age)
print(type(age))
```

```python
age =15 # int
age= list(age)
print(age)
print(type(age))
```

```python
# from string to tuple
age="25"
age=tuple(age)
print(age)
print(type(age))
```

```python
# from int to tuple
age=25
age=tuple(age)
print(age)
print(type(age))
```

```python
age ="25"
age=set(age)
print(age)
print(type(age))
```

```python
age="hello"
age=dict(age)
print(age)
print(type(age))
```

```python
## Why Type conversion is required
price = float(input("Enter product price: ")) # giving input as price #input function always return string
quantity = 2
total = price * quantity
print("Total price:", total)
```

## fstring - Format string

### It was introduced in python 3.6

```python
name = "chitti"
print("Hello", name)
print("Hello " + name)
print(f"Hello {name}") # fstring
```
