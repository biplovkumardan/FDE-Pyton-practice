```python
# 2. Connect to MongoDB
# 'MongoClient' connects Python to the MongoDB server running on this computer
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
print("Connected to MongoDB!")
```

```python
# 3. Select database and collection
# 'FDE' is DB
db = client["FDE"] 
students = db["students"]   # students is a collection

print("Database and collection selected!")
```

```python
# 4. CREATE - Add a student
student = {"name":"BKD", "age":22, "course":"FDE-Database-mongo"}

students.insert_one(student)
print("Student Added!")
```

```python
# 5. READ - show all students
for student in students.find():
    #print(f"{student["name"]}, {student["age"]},{student["course"]}")
    print(student["name"],student["age"],student["course"])
```

```python
result = students.update_one(
    {"name": "Veera"},
    {"$set":{"age": 27}}
)
print("Matched:", result.matched_count)
print("Modified:", result.modified_count)
```
