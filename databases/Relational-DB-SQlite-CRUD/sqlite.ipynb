## Sqlite (Relational DB service) CRUD Operation Python

### SQLite is light weight server filebased database ##
### Sqlite is available through the built-in sqlite3 feature in python ###
### Not required to install separate database server like MySQL

```python
import sqlite3
# sqlite3 is built-in python module
# it provides functionalities to communicate with SQLite Database
```

```python
## DB connection ##
conn = sqlite3.connect("dev.db")
print("Database connected successfully")
# if dev.db not exists it will create automatically
# connect() is a function
# conn is variable. It stores connection object
```

```python
cur = conn.cursor()
# cursor() is instance method and creates object
# cur is variable
# cursor purpose is to execute SQL queries
```

```python
cur.execute("""
    CREATE TABLE students (
        id INTEGER PRIMARY KEY,
        name TEXT,
        age INTEGER,
        course TEXT
    )
""")
conn.commit()

# Execute SQL statement
# commit() save changes permanently
```

```python
# cur.execute("""
#     DROP TABLE students
# """)
# conn.commit()
```

```python
# Static query execution
cur.execute("""
    INSERT INTO students (name, age, course)
    VALUES ("Biplov Dan", 25, "FDE")
""")
conn.commit()
```

```python
# Dynamic query execution
query = """
    INSERT INTO students (name, age, course)
    VALUES (?, ?, ?)
"""
name = "BiplovD"
age = 26
course ="FDE-DB"
cur.execute(query, (name, age, course))
conn.commit()
```

```python
cur.execute("SELECT * FROM students")
records = cur.fetchall()
print(records)
```
