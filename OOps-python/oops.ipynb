```python
# OOps has 2 types of variables. 1. Class variable 2. Instance variable
class Student: # class is python keyword. # Student is my class name
    name = "John" # Class variable
    age = 20 # Class variable
```

```python
# Instance variable is a variable that is defined inside a method and is associated with an instance of the class. 
# example
class Student:
    def __init__(self, name, age):
        self.name = name # instance variable
        self.age = age
```

## init method
__init__ constructor

```python
# If __init__ method is not defined in the class, then the default constructor is called. 
# The default constructor does not take any parameters and does not perform any operations. It simply creates an instance of the class.
class Student:
    country = "India"   # class variable
    def __init__(self, name, age):
        self.name = name # instance variable
        self.age = age

s1 = Student("John", 20) # creating an instance of the class
print(s1.name) # accessing instance variable
print(s1.age) # accessing instance variable
print(s1.country) # accessing class variable
```

```python
# if dispaly method is defined in the class, then the default constructor is not called. 
# Check the difference between the two classes. In the first class, the default constructor is called and the instance variables are not initialized. 
# In the second class, the __init__ method is defined and the instance variables are initialized.
# The display method takes two parameters, name and age, and assigns them to the instance variables self.name and self.age.
class Student:
    country = "India"   # class variable
    def display(self, name, age):
        self.name = name # instance variable
        self.age = age

s1 = Student() # creating an instance of the class
s1.display("John", 20) # calling the display method
print(s1.name) # accessing instance variable
print(s1.age) # accessing instance variable
print(s1.country) # accessing class variable with refer s1 variable
print(Student.country) # accessing class variable with refer class name
```
