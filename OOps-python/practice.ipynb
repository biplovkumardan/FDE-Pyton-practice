## OOPs: Object Oriented Programming

### OOPs is a programming approach where we organise our code using Classes and Objects

### Class is blueprint or template to create objects


```python
# Basic Syntax of Class and Object in Python
class ClassName: # class is python keyword. # ClassName is my class name
    class_variable1 = "value1" # Class variable
    class_variable2 = "value2"
    class_variable3 = "value3"

c1 = ClassName() # () creating object of class ClassName. c1 is object of class ClassName
```

```python
class Car: # class is python keyword. # Car is my class name
    name = "Skoda" # Class variable
    color = "Red"
    speed =100
c1 = Car() # () creating object of class Car. c1 is object of class Car
print(c1.name)
print(c1.color)
print(c1.speed)


# c1 --> (name = "Skoda", color = "Red") # c1 is object of class Car. It has two attributes name and color
```

```python
class Student: # class is python keyword. # Student is my class name
    name = "Biplov" # Class variable
    roll = 235
    age= 15
c1 = Student() # () creating object of class Student. c1 is object of class Student
print(c1.name)
print(c1.roll)
print(c1.age)
```

```python
# with function example
class Car: # class is python keyword. # Car is my class name
    name = "Skoda" # Class variable
    color = "Red"
    speed =100
    def display(self): # function inside class is called method. self is a reference to the current instance of the class.
        print(f"Car Name: {self.name}, Color: {self.color}, Speed: {self.speed}")
        #return f"Car Name: {self.name}, Color: {self.color}, Speed: {self.speed}"

c1 = Car() # () creating object of class Car. c1 is object of class Car
c1.display() # calling method display of class Car using object c1  
```

```python
class Car:
    def display(self):
        print(self)

c1 = Car()
print(c1) # printing object c1 of class Car
c1.display() # calling method display of class Car using object c1

# we got same response for both print(c1) and c1.display() because self is a reference to the current instance of the class. 
# So, when we call c1.display(), self refers to c1, and when we print(c1), it also refers to the same object c1.
```

```python
# with oops example for Car1 and Car2
class Car:
    def __init__(self, color, speed, fuel): # constructor method to initialize the object
        self.color = color
        self.speed = speed
        self.fuel = fuel

    def start(self):
        print(f"Car is starting with color {self.color}, speed {self.speed} and fuel {self.fuel}")

    def stop(self):
        print(f"Car is stopping with color {self.color}, speed {self.speed} and fuel {self.fuel}")

    def move(self):
        print(f"Car is moving with color {self.color}, speed {self.speed} and fuel {self.fuel}")
```

```python
car1 = Car("Red", 100, "Petrol")
car2 = Car("Blue", 120, "Diesel")
car3 = Car("Green", 150, "Electric")
car1.start()
car1.stop()
car1.move()
car2.start()
car2.stop()
car2.move()
car3.start()
car3.stop()
car3.move()
```
