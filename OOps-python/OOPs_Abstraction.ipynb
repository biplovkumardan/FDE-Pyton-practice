# Abstraction

### Abstraction class define WHAT should be done, and Child class defines HOW it should be done

```python
from abc import ABC, abstractmethod

class ATM(ABC): # for understanding consider parent class as ATM and child class as SBI, HDFC, ICICI etc
    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def check_balance(self):
        pass
class SBI(ATM):
    def __init__(self):
        self.balance = 10000
    
    def withdraw(self, amount):
        self.balance -= amount
        print(f"Withdrawn {amount} from SBI account. New balance: {self.balance}")

    def check_balance(self):
        print(f"Current balance in SBI account: {self.balance}")

class HDFC(ATM):
    def __init__(self):
        self.balance = 20000
    
    def withdraw(self, amount):
        self.balance -= amount
        print(f"Withdrawn {amount} from HDFC account. New balance: {self.balance}")

    def check_balance(self):
        print(f"Current balance in HDFC account: {self.balance}")

# create objects of SBI and HDFC classes
s = SBI()
h = HDFC()

s.withdraw(2000)
s.check_balance()

h.withdraw(5000)
h.check_balance()
```
