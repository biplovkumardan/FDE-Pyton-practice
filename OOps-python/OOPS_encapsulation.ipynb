# Encapsulation

### Encapsulation helps protect an object's data from accidental direct modification by providing controlled ways to access or modify it


```python
# without encapsulation
class Bank:
    bank_name = "HDFC" # class variable
    def __init__(self, name, balance):
        self.name = name # instance variable
        self.balance = balance # instance variable

    def show_balance(self):
        print(f"{self.name} has a balance of {self.balance} in {self.bank_name}")

    def deposit(self, amount):
        self.balance += amount
        print(f"{self.name} has deposited {amount}. New balance is {self.balance}")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print(f"{self.name} has insufficient balance to withdraw {amount}. Current balance is {self.balance}")
        else:
            self.balance -= amount
            print(f"{self.name} has withdrawn {amount}. New balance is {self.balance}")

b1 = Bank("John", 10000)

b1.show_balance()  
b1.deposit(5000)
b1.show_balance()  
b1.withdraw(16000)
b1.balance = 0 # John can directly change the balance without using the withdraw method. 
#This is a problem because it allows John to bypass the withdraw method and set his balance to 0, which is not allowed.
# This is an example of how encapsulation can help prevent this kind of problem by restricting access to the balance variable.
b1.show_balance()  
```

```python
# with encapsulation
#----------------------------------------
# Protect data inside the class by making it private. This is done by adding two underscores before the variable name.
# we can control the access to the private variable by providing public methods to access and modify the private variable. This is called encapsulation.
class Bank:
    bank_name = "HDFC" # class variable
    def __init__(self, name, balance):
        self.name = name # instance variable
        self.__balance = balance # private instance variable. No one can access this variable directly from outside the class. It can only be accessed through the methods of the class.

    def show_balance(self):
        print(f"{self.name} has a balance of {self.__balance} in {self.bank_name}")

    def deposit(self, amount):
        self.__balance += amount
        print(f"{self.name} has deposited {amount}. New balance is {self.__balance}")
    
    def withdraw(self, amount):
        if amount > self.__balance:
            print(f"{self.name} has insufficient balance to withdraw {amount}. Current balance is {self.__balance}")
        else:
            self.__balance -= amount
            print(f"{self.name} has withdrawn {amount}. New balance is {self.__balance}")

b1 = Bank("John", 10000)

b1.show_balance()  
b1.deposit(5000)
b1.show_balance()  
b1.withdraw(14000)
b1.__balance = 0 # even if John tries to change the balance directly, it will not work because __balance is a private variable and cannot be accessed from outside the class.
b1.show_balance()  
```
