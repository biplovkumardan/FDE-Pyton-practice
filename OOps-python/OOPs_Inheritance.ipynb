# Inheritance

### Inheritance method is an oops concept that allows one class to acquire the properties and methods of another class

```python
# Without inheritance
# Separate classes Parent and Child
class Parent: #create class
    def house(self): #create method
        print("Parent own house") #print statement
    def car(self): #create method
        print("Parent own car") #print statement

class Child: #create class
    def bike(self): #create method
        print("Child own bike") #print statement
    def cycle(self): #create method
        print("Child own cycle") #print statement

p1 = Parent() #create object of class Parent
p1.house() #call method house of class Parent
p1.car() #call method car of class Parent

c1 = Child() #create object of class Child
c1.bike() #call method bike of class Child
c1.cycle() #call method cycle of class Child
```

```python
# With inheritance
class Parent:
    def house(self):
        print("Parent own house")
    def car(self):
        print("Parent own car")

class Child(Parent): #create class Child that inherits from Parent
    def bike(self):
        print("Child own bike")
    def cycle(self):
        print("Child own cycle")

# p1 = Parent() #create object of class Parent
# p1.house() #call method house of class Parent
# p1.car() #call method car of class Parent

c1= Child() #create object of class Child
## Trying to access parent method from child method using inheritance
c1.house() #call method house of class Parent from Child object
c1.car() #call method car of class Parent from Child object
## trying to access child method 
c1.bike() #call method bike of class Child
c1.cycle() #call method cycle of class Child

# Note: Not created any object for parent but still able to call parent method using child() object with the help of inheritance.
```

```python
# Inheritance example - 2
class Bank:

    def __init__(self):
        self.__balance = 1000 #private variable

    def show_balance(self):
        print("Available Balance:", self.__balance)

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

class SavingsAccount(Bank): #create class SavingsAccount that inherits from Bank
    def interest(self):
        print("Savings account interest: 5%")

class CurrentAccount(Bank): #create class CurrentAccount that inherits from Bank
    def business(self):
        print("Business transactions enabled")

# savings Account: for saving account I am using bank() class using inheritance
s1 = SavingsAccount() #create object of class SavingsAccount

s1.show_balance() #call method show_balance of class Bank from SavingsAccount object
s1.deposit(5000)
s1.show_balance() #call method show_balance of class Bank from SavingsAccount object
s1.interest() #call method interest of class SavingsAccount

print("**"*20)

# current Account: for current account I am using bank() class using inheritance
c1 = CurrentAccount() #create object of class CurrentAccount

c1.show_balance() #call method show_balance of class Bank from CurrentAccount object
c1.deposit(2000)
c1.show_balance() #call method show_balance of class Bank from CurrentAccount object
c1.business() #call method business of class CurrentAccount

```

## super method

```python
class Parent:
    def display(self):
        print("Parent class method")

class Child(Parent): #inherit from Parent class
    def display(self):
        print("Child class method")
        super().display()
c1 = Child() #create object of class Child
c1.display() #call method display of class Child
```

```python
# Inheritance with super init
class bank:
    def __init__(self):
        self.balance = 1000 

class SavingsAccount(bank): #create class SavingsAccount that inherits from bank
    def __init__(self):
        super().__init__() #call parent class constructor
        self.account_type = "Savings Account" #public variable

s1 = SavingsAccount() #create object of class SavingsAccount
print(s1.account_type) #print account_type of class SavingsAccount
print(s1.balance) #print balance of class SavingsAccount
```
