# Polymorphism

## Same method name, but we can perform multiple operations

```python
class Payment:
    def pay(self, amount):
        print(f"Processing payment of {amount} made.")

class GooglePay(Payment):
    def pay(self, amount):
        print(f"Processing Google Pay payment of {amount} made.")

class PhonePe(Payment):
    def pay(self, amount):
        print(f"Processing PhonePe payment of {amount} made.")

class CreditCard(Payment):
    def pay(self, amount):
        print(f"Processing Credit Card payment of {amount} made.")

class CashOnDelivery(Payment):
    def pay(self, amount):
        print(f"Processing Cash on Delivery payment of {amount} made.")

# Customer selects different payment methods
payment_method = input("Select payment method (GooglePay/PhonePe/CreditCard/CashOnDelivery): ")

if payment_method == "GooglePay":
    g = GooglePay()
    g.pay(1000)
elif payment_method == "PhonePe":
    p = PhonePe()
    p.pay(2000)
elif payment_method == "CreditCard":
    c = CreditCard()
    c.pay(3000)
elif payment_method == "CashOnDelivery":
    cod = CashOnDelivery()
    cod.pay(4000)


```
