## Indentation and systax same process follow as if

```python
# for loop we can use iterate over sequence of items, list, tupel, range, dictionary etc
# for loop with string example
a ="python"
for i in a:
    print(i)
```

```python
#for loop with break
number = [1, 2, 3, 4, 5, 6]
for i in number:
    if i == 2:
        break
    print(i)
```

```python
# for loop with continue
number = [1, 2, 3, 4, 5, 6]
for i in number:
    if i == 2:
        continue
    print(i)
# skip loop for 2, other loop will run
```

```python
## for loop with dictionary
## for loop calling values
number = {
    "name" : "biplov",
    "course" : "FDE"
}
for i in number.keys():
    print(i)
```

```python
number = [10, 20, 30, 40, 50]
for i in number:
    if i == 40:
        print("Number found")
        break
    print(i)
```

```python
# for loop with range example
total = 0
dev = range(1,6)
for i in dev:
    total = total+i
print("Total:", total)
# return = 1 + 2 + 3 + 4 + 5 i.e in range(1,6) sums up all except the last one
```

```python
# random
import random
for i in range(1,5):
    num = random.randint(1,200)
    if num%3==2:
        print("Number found", num)
        break
    else:
        print("Try again", num)
```
