### Incremental Decremental Operators

```python
print("Incremental Operators")
# In Python, we can achieve incrementing by using Python '+=' operator. There is nothing like '++' operator
x =13
x+=5
print(x)
print("INCREMENTED FOR LOOP")
for i in range(0, 5):
    print(i)

print()
print( "Decremental Operators")
# Python does not have a specific decrement operator in Python (like -- in some other programming languages). 
# However, you can achieve decrementing a variable using the -= operator.
z=15
z-=7
print(z)

print("DECREMENTED FOR LOOP")
for i in range(4, -1, -1):
    print(i)
```

## Bitwise Operators

### Left Shift (<<) and Right Shift (>>)

```python
#Left Shift (<<): Shift left by pushing zeros in from the right 
a = 16
a << 1  # calculates the shifted value, but it does not save or reassign that new value back to a.
print(a)  # value of a will remain the same


x = 15  # Binary of 15 is 0000 1111.
x <<= 1  #left shift (<<) : move bits left and add bit postion on right 0. As it is shift left = 1 add one 0.
# x will become in binary format 0000 1111 <<= 1 to 0001 1110
# the decimal format of 0001 1110 is 30
print(x)
```

```python
x=19  # Binary: 0001 0011
x<<=2 # Binary after 2 Left shift: 0100 1100
print(x) # pow(2,6) + pow(2.3) + pow(2, 2) = 64+8+4 = 76

# Here i am used pow only for my own understanding of power on 2.
# In pow(2, p), p is the position of 1 from right-most binary digit. Right most binary postion always starts with 0.
```

```python
# Right Shift (>>): Shifts the bits of the number to the right and fills the vacated bit positions on the left with 0s
x=15 # Binary: 1111
x>>=1 # Becomes: 0111. That is 7 in decimal format
print(x)
```

```python
x=15
x>>=4
print(x)
```

```python
x=15
x>>=5
print(x)
```

### Bitwise AND (&)

```python
# Example ( either of 1 then 0, Both of 1 then 1 and Both of 0 then 0)
a = 5
b = 3
print("Bitwise AND:", a & b)

# a= 5: binary is 0101
# b= 3: binary is 0011
# 0101 & 0011 = 0001. Decimal of 0001 is 1
```

```python
# task
a=8  # Binary: 1000
b=4  # Binary: 0100
# & operator:  0000
print("Bitwise AND:", a & b) # Decimal of 0000 is 0
```

### Bitwise OR (|)

```python
# example ( either of 1 then 1, Both of 1 then 1 and Both of 0 then 0)
a=5 #  binary is 0101
b=3 # binary is 0011
# 0101 | 0011 = 0111. Decimal of 0111 is 7
print("bitwise OR:", a | b)
```

### Bitwise XOR (^)

```python
# Example:  either of 1 then 1, Both of 1 then 0 and Both of 0 then 0
# The ^ operator compares each bit and set it to 1 if only one is 1, otherwise (if both are 1 or both are 0) it is set to 0
a=5 #  binary is 0101
b=3 # binary is 0011
# 0101 ^ 0011 = 0110. Decimal is 6
print("Bitwsie operator:", a ^ b)
```

### Bitwise Not (~)

```python
# Example
a = 5 # Binary 0101
# ~a = -(a+1) IMPORTANT RULE to KEEP in the MIND
# that is ~5 = -(5+1) =-6
print(~a)

## UNDER the hood
# Binary of 5: In an 8-bit representation, 5 looks like this: 00000101
# Inverting the bits (~): Flipping every 0 to 1 and 1 to 0 gives: 11111010
# Interpreting as a negative number: Because the leftmost bit is 1, 
# the computer treats it as a negative number in two's complement form, which evaluates to -6.
```
