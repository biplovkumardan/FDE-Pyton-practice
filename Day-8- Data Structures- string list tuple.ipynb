### 1.1 String Indexing
#### Indexing: We can call individual letter from word at a particular position

####   word            = " P   Y   T   H   O   N "
####   Positive index      0   1   2   3   4   5
####   Negative index     -6  -5  -4  -3  -2  -1    
-----------------------------------------------------------
#### Positive index start from 0 (left) ----->
#### Negative index start from -1 (right) <------

```python
# Positive index
word = "PYTHON"
print(word[0])
print(word[3])
print(word[5])
```

```python
# Negative index
print(word[-6])
print(word[-3])
print("Last character of the string is:", word[-1])
```

```python
fruits = ["orange", "grape", "banana", "kiwi"]
print(fruits[0])
print(fruits[-2])
```

```python
print(fruits[4])
```

## 1.2 String Slicing
### Slicing means extracting a portion of string

#### start: index where slicing begins, included
#### stop: index where slicing ends, excluded
#### step: number of postions to shift. if +ve from left to right and if -ve from right to left
##### Positive Step: start should be less than end (left to right) incremental
##### Negative Step: start should be greater than stop (right to left) decremental

### Ex: [start : stop : step]

## Positive Slicing Process

```python
word ="PYTHON"
print(word[3:5])
# if step is blank, it will take default value 1
```

```python
word ="PYTHON"
print(word[5:3])
# the result will be empty because the start index is greater than the end index and the step is positive.
```

```python
word ="PYTHON"
print(word[3:9])
# if the stop index is greater than the length of the string, it will take the length of the string as the stop index.
```

```python
# Omitting the start index
print(word[:3])
# if start index is blank, it will take default value 0
```

```python
# Omitting the stop index
print(word[3:])
# if stop index is blank, it will take default value length of the string
```

```python
# omitting both start and stop index
print(word[:])
# if both start and stop index are blank, it will take default value 0 and length of the string respectively.
```

## Negative Slicing Process

```python
word ="PYTHON"
print(word[-4:3])
# start index -4 has char T (included) and stop index 3 has char H (excluded) and step is postive, so the result is T only
```

```python
print(word[-4:2])
# start index -4 has char T (included) and stop index 2 has char T (excluded) and step is postive, so the result is blank only
```

```python
word ="PYTHON"
print("Result1:", word[-4:-5])
# the Result1 is blank because -4 is greater than -5 but for +ve step it should be less than -5
print("Result2:", word[-6:-5])
```

```python
word ="PYTHON"
print("Result1:", word[-4:-6:-1])
# the Result1 is T because -4 is greater than -6 and for -ve step it should move from right to left
```

```python
word ="PYTHON"
print("Result1:", word[7:1:-2])
```

```python
word = "Biplov Kumar Dan"
print(word[1:6:2])
```

```python
word = "Biplov Kumar Dan"
print(word[9:0:-2])
```
