# Day-1 Identifiers and Variables
### Identifier: A name used to identify a program element such as variables, functions, class etc..
### Variables: A name that refer to an object in memory during execution

## Rules to define a Variable or Identifier

#### Use letter and digits and underscore_
#### Never start Identifiers with digit
#### Don't use python keywords as Identifiers ex: if, for etc..
#### Do not use space or hyphens - in Identifier
#### Better to use snake_case ex server_name, SERVER_NAME
#### Avoid using python built-in functions name as variable name

### Keyword: Python Keyword list

```python
import keyword
a=keyword.kwlist
print(a)
print(len(a))
```

### Do not use keywords as variables

```python
# Here 'if' is a keyword used as variable
if=7
print(if)
```

### Built-in: Built-in are the functions ready to use (already defined functions)

```python
import builtins
a=dir(builtins)
print(a)
print(len(a))
```

### Builtins can be used as variable but we should avoid it

```python
# Builtins example
len=9
print(len)
#when builtins are used, no error but better to avoid
```

### Builtin if used as variable can create conflict

```python
len=5
print(len)
print(len("Biplov"))
```

### Test a word is keyword or not

```python
import keyword
print(keyword.iskeyword("if"))
print(keyword.iskeyword("len"))
```

### Test a word is builtins or not

```python
import builtins
print("print" in dir(builtins))
print("if" in dir(builtins))
```

## Recap Below Rules for Defining Variables and Identifiers in Python

### Rules for Defining Variables and Identifiers in Python

#### 1. Use letters, digits and Underscores('_')
#### Identifiers can contain letters ('A-Z', 'a-z'), digits('0-9'), and underscores ('_').

```python
## Examples:

server_name ="localhost"
server123 ="testServer"
_serverEnvironment ="production"
```

#### 2. Never start an identifier with a digit
#### An identifier can not begin with a digit.

```python
## Valid
Server1 ="Localhost"
OS1 = "Windows"
```

```python
## Invalid
1Server ="localhost"
6OS="Linux"
```

#### 3. Do not use Python keywords as identifiers
#### Python keywords such as 'if', 'for', 'while', 'class' , 'def' etc. cannot be used as variable names

```python
## Invalid
if=5
for =45
```

```python
## Valid
if_name ="biplov"
for_count= 8
```

#### 4. Do not use spaces or hyphens ('-') in identifiers
#### Identifiers cannot contain spaces or hyphens. [Identifiers: variables, Class, function and other programming elements]

```python
# Invalid
def sumof twoNumber(): #space in function identifier
 sum-of-number=2 # hyphen in variable identifier
```

#### 5. Prefer 'snake_case' for variable and function names
#### Python's PEP 8 style guide recommends 'snake_case' for variables and functions

```python
# Example
'''
Component	                 Convention	                        Example
---------------------------------------------------------------------------------------------
Variables & Functions	     snake_case	                        user_count, calculate_total()
Classes	                     PascalCase	                        UserProfile, ShoppingCart
Constants	                 UPPER_SNAKE_CASE	                MAX_CONNECTIONS, DEFAULT_TIMEOUT
Modules & Packages	         snake_case (short, lowercase)	    utils.py, data_parser
Private Attributes/Methods	 Leading underscore (_snake_case)	_internal_cache
'''
```

#### 6. Avoid using Python built-in names as variable names
#### Avoid naming variables after built-in functions, types, or objects, such as 'print', 'list', 'str' and 'sum', because doing so shadows the original built-in

```python
# Not Recommended

print=200
list =[1,2,3]

# After assigning these names , you may be unable to use the corresponding built-ins normally within that scope.

## Recommended
print_value= 1234
list_number =[1,4,7,9]
```
